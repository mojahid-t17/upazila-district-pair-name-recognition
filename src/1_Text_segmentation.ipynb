```python
# =========================================================================
# === 1.  SETUP, IMPORTS, FUNCTIONS, AND INITIAL BOX DETECTION ===
# =========================================================================

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import ipywidgets as widgets
import math
# pytesseract is imported for completeness, though not actively used for output here
import pytesseract
from google.colab.patches import cv2_imshow
```

```python
# ================================
# 2. TEXT REGION DETECTION
# ================================
def detect_text_regions(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # MSER detection
    mser = cv2.MSER_create(5, 150, 10000)
    regions, _ = mser.detectRegions(gray)
    mser_boxes = []
    for p in regions:
        x, y, w, h = cv2.boundingRect(p)
        if 10 < w < 800 and 10 < h < 250:
            mser_boxes.append((x, y, x+w, y+h))

    # Canny + dilation
    edges = cv2.Canny(gray, 50, 150)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (9, 3))
    dilated = cv2.dilate(edges, kernel, 2)

    cnts, _ = cv2.findContours(dilated, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    canny_boxes = []
    for c in cnts:
        x, y, w, h = cv2.boundingRect(c)
        if 15 < w < 900 and 12 < h < 250:
            canny_boxes.append((x, y, x+w, y+h))

    return mser_boxes + canny_boxes
```

```python

# ================================
# 3. COLUMN SPLIT DETECTION
# ================================
def detect_column_split(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    th = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
                               cv2.THRESH_BINARY_INV, 41, 5)
    vertical_proj = np.sum(th, axis=0)
    h, w = gray.shape
    mid1 = w // 3
    mid2 = 2 * w // 3
    idx = np.argmin(vertical_proj[mid1:mid2])
    return mid1 + idx
```

```python
# ================================
# 4. ROW GROUPING
# ================================
def group_boxes_into_rows(boxes, row_height=90):
    boxes_sorted = sorted(boxes, key=lambda b: b[1])
    rows = []
    for b in boxes_sorted:
        bx1, by1, bx2, by2 = b
        cy = (by1 + by2) // 2
        placed = False
        for row in rows:
            avg_y = np.mean([(r[1] + r[3]) // 2 for r in row])
            if abs(cy - avg_y) <= row_height:
                row.append(b)
                placed = True
                break
        if not placed:
            rows.append([b])
    return rows
```

```python
# ================================
# 5. MERGE BOXES USING MINAREARECT
# ================================
def merge_adjacent_boxes_with_minAreaRect(boxes, col_split, is_left, gap=35):
    if not boxes:
        return []
    boxes = sorted(boxes, key=lambda b: b[0])
    merged_boxes = []
    current_group = [boxes[0]]
    current_box = boxes[0]

    for b in boxes[1:]:
        x1, y1, x2, y2 = b
        cx1, cy1, cx2, cy2 = current_box
        horizontal_gap = x1 - cx2
        vert_overlap = max(0, min(cy2, y2) - max(cy1, y1))
        min_height = min(cy2 - cy1, y2 - y1)
        should_merge = (horizontal_gap <= gap or
                        (vert_overlap > 0 and (vert_overlap / min_height) > 0.45))
        if should_merge:
            nx2 = max(cx2, x2)
            if (is_left and nx2 > col_split) or (not is_left and nx2 < col_split):
                should_merge = False
        if should_merge:
            current_box = (min(cx1, x1), min(cy1, y1), max(cx2, x2), max(cy2, y2))
            current_group.append(b)
        else:
            points = []
            for box_ in current_group:
                x1b, y1b, x2b, y2b = box_
                points.extend([[x1b, y1b], [x2b, y1b], [x1b, y2b], [x2b, y2b]])
            points = np.array(points, dtype=np.float32)
            if len(points) >= 3:
                rect = cv2.minAreaRect(points)
                center, (w, h), angle = rect
                x = center[0] - w / 2
                y = center[1] - h / 2
                merged_boxes.append((x, y, w, h, angle))
            current_box = b
            current_group = [b]

    # Final group merge
    if current_group:
        points = []
        for box_ in current_group:
            x1b, y1b, x2b, y2b = box_
            points.extend([[x1b, y1b], [x2b, y1b], [x1b, y2b], [x2b, y2b]])
        points = np.array(points, dtype=np.float32)
        if len(points) >= 3:
            rect = cv2.minAreaRect(points)
            center, (w, h), angle = rect
            x = center[0] - w / 2
            y = center[1] - h / 2
            merged_boxes.append((x, y, w, h, angle))

    return merged_boxes
```

```python
# ================================
# 6. FULL DETECTION + CROPPING
# ================================
def detect_and_crop_boxes(img):
    raw_boxes = detect_text_regions(img)
    col_split = detect_column_split(img)
    left, right = [], []
    for b in raw_boxes:
        x1, y1, x2, y2 = b
        cx = (x1 + x2) // 2
        if cx < col_split:
            left.append(b)
        else:
            right.append(b)
    left_rows = group_boxes_into_rows(left)
    right_rows = group_boxes_into_rows(right)
    final_boxes = []
    for r in left_rows:
        final_boxes.extend(merge_adjacent_boxes_with_minAreaRect(r, col_split, True))
    for r in right_rows:
        final_boxes.extend(merge_adjacent_boxes_with_minAreaRect(r, col_split, False))

    # Crop rotated boxes
    cropped = []
    for (x, y, w, h, angle) in final_boxes:
        rect_center = (x + w/2, y + h/2)
        M = cv2.getRotationMatrix2D(rect_center, angle, 1.0)
        rotated = cv2.warpAffine(img, M, (img.shape[1], img.shape[0]),
                                 flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_REPLICATE)
        x1 = int(rect_center[0] - w/2)
        y1 = int(rect_center[1] - h/2)
        x2 = int(rect_center[0] + w/2)
        y2 = int(rect_center[1] + h/2)
        crop = rotated[max(0,y1):max(0,y2), max(0,x1):max(0,x2)]
        if crop.size > 0:
            cropped.append(crop)

    return final_boxes, col_split, cropped
```

```python
# ================================
# 7. SHOW EDGE DETECTION
# ================================
def show_edge_detection(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)
    dilated = cv2.dilate(edges, cv2.getStructuringElement(cv2.MORPH_RECT, (9, 3)), 2)
    fig = plt.figure(figsize=(15, 5))
    plt.subplot(1,3,1)
    plt.title("Gray")
    plt.imshow(gray, cmap="gray")
    plt.axis("off")
    plt.subplot(1,3,2)
    plt.title("Edges")
    plt.imshow(edges, cmap="gray")
    plt.axis("off")
    plt.subplot(1,3,3)
    plt.title("Dilated")
    plt.imshow(dilated, cmap="gray")
    plt.axis("off")
    plt.show()
```
